## Maimal Square

```python
class Solution:
    def maximalSquare(self, matrix: List[List[str]]) -> int:
        maxLen = 0
        m=len(matrix)
        n=len(matrix[0])
        dp={}
        def help(i,j):
            if i>=m or j>=n:
                return 0
            if matrix[i][j]=="0":
                return 0
            if (i,j) in dp:
                return dp[(i,j)]
            dp[(i,j)] =  1 + min(help(i+1, j), help(i, j+1), help(i+1, j+1))
            return dp[(i,j)]
        ma = float('-inf')
        for i in range(m):
            for j in range(n):
                if matrix[i][j]=="1":
                    curr = help(i, j)
                    ma=max(curr, ma)
        return ma*ma if ma!=float('-inf') else 0
```

## Count Square Submatrices with All Ones

```python
class Solution:
    def countSquares(self, matrix: List[List[int]]) -> int:
        m=len(matrix)
        n=len(matrix[0])
        dp={}
        def help(i,j):
            if i>=m or j>=n:
                return 0
            if matrix[i][j]==0:
                return 0
            if (i,j) in dp:
                return dp[(i,j)]
            dp[(i,j)] = 1 + min(help(i+1,j),help(i, j+1),help(i+1, j+1))
            return dp[(i,j)]
        s=0
        for i in range(m):
            for j in range(n):
                s+=help(i,j)
        return s
```

## Minimum Falling Path Sum

```python
class Solution:
    def minFallingPathSum(self, matrix: List[List[int]]) -> int:
        m=len(matrix)
        n=len(matrix[0])
        dp={}
        def dfs(i,j):
            if  j<0 or j>=n:
                return float('inf')
            if i==m-1:
                return matrix[i][j]
            if (i,j) in dp:
                return dp[(i,j)]
            dp[(i,j)] = matrix[i][j]+min(dfs(i+1, j-1), dfs(i+1, j), dfs(i+1, j+1))
            return dp[(i,j)]
        mi = float('inf')
        for i in range(m):
            curr = dfs(0, i)
            mi=min(curr, mi)
        return mi
```
