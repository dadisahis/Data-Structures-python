## Unique Paths

```python
class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        #recurrence
        def help(c, i, j):
            if i==0 and j==0:
                c+=1
                return c
            if i<0 or j<0:
                return 0

            c += help(c,i-1, j ) + help(c, i, j-1)
            return c
        
        return help(0,m-1,n-1)

# TC: 2^(m+n) SC: O(m+n)


class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        #top down memoization
        dp = [[-1 for i in range(n)] for j in range(m)]
        def help(c, i, j):
            if i==0 and j==0:
                c+=1
                dp[i][j]=c
                return dp[i][j]
            if i<0 or j<0:
                return 0

            if dp[i][j]!=-1:
                return dp[i][j]
            c += help(c,i-1, j ) + help(c, i, j-1)
            dp[i][j]=c
            return dp[i][j]
        help(0,m-1,n-1)
        return dp[-1][-1]

# TC - O(m*n) SC- O(m*n) + O(m+n) - O(m+n) is recursion stack space

class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        dp = [[0 for i in range(n)] for j in range(m)]
        dp[0][0]=1
        for i in range(n):
            dp[0][i]=1
        for i in range(m):
            dp[i][0]=1
        for i in range(1, m):
            for j in range(1, n):
                dp[i][j] += dp[i-1][j]+dp[i][j-1]
        return dp[-1][-1]

# TC - O(m*n) SC- O(m*n) this saves recursive stack
```

## Maimal Square

```python
class Solution:
    def maximalSquare(self, matrix: List[List[str]]) -> int:
        maxLen = 0
        m=len(matrix)
        n=len(matrix[0])
        dp={}
        def help(i,j):
            if i>=m or j>=n:
                return 0
            if matrix[i][j]=="0":
                return 0
            if (i,j) in dp:
                return dp[(i,j)]
            dp[(i,j)] =  1 + min(help(i+1, j), help(i, j+1), help(i+1, j+1))
            return dp[(i,j)]
        ma = float('-inf')
        for i in range(m):
            for j in range(n):
                if matrix[i][j]=="1":
                    curr = help(i, j)
                    ma=max(curr, ma)
        return ma*ma if ma!=float('-inf') else 0
```
